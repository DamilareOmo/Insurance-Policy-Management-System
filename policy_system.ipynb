# 🛡️ Insurance Policy Management System
A demonstration of the **Policyholder**, **Product**, and **Payment** classes.  
Run each cell in order using **Shift + Enter**.

---

## 1. Import Classes

```python
from policyholder import Policyholder
from product import Product
from payment import Payment
from csv_exporter import CSVExporter

print("✅ All classes imported successfully.")
```

## 2. Create Insurance Products

```python
health_plan = Product(
    product_id       = "P001",
    name             = "Family Health Plan",
    premium          = 150.00,
    coverage_details = "Covers hospitalization, outpatient care, and dental."
)

auto_plan = Product(
    product_id       = "P002",
    name             = "Comprehensive Auto Plan",
    premium          = 90.00,
    coverage_details = "Covers collision, theft, and third-party liability."
)

print("Products created:")
health_plan.display_details()
auto_plan.display_details()
```

## 3. Update a Product

```python
auto_plan.update(premium=95.00)
print("\nUpdated Auto Plan:")
auto_plan.display_details()
```

## 4. Register Policyholders

```python
holder1 = Policyholder("M001", "Amaka Obi",    "amaka.obi@example.com",  "555-1010")
holder2 = Policyholder("M002", "John Mensah",  "john.mensah@example.com","555-2020")

holder1.register()
holder2.register()

holder1.assign_product(health_plan)
holder2.assign_product(auto_plan)
```

## 5. Process Payments

```python
payment1 = Payment("PMT001", holder1, health_plan, health_plan.premium)
payment2 = Payment("PMT002", holder2, auto_plan,   auto_plan.premium)

# Holder 1 — pays in full
payment1.process_payment(150.00)

# Holder 2 — partial payment first
payment2.process_payment(50.00)
```

## 6. Send a Payment Reminder

```python
payment2.send_reminder()

# Holder 2 completes their payment
payment2.process_payment(45.00)
```

## 7. Apply a Late Payment Penalty

```python
payment3 = Payment("PMT003", holder1, health_plan, health_plan.premium)
payment3.send_reminder()
payment3.apply_penalty()
payment3.process_payment(157.50)   # pays full amount including 5% penalty
```

## 8. Suspend & Reactivate a Policyholder

```python
holder2.suspend()
holder2.reactivate()
```

## 9. Suspend a Product

```python
auto_plan.suspend()
auto_plan.display_details()
```

## 10. Display Final Account Details

```python
print("\n" + "="*50)
print("ACCOUNT DETAILS — AMAKA OBI")
print("="*50)
holder1.display_details()

print("\n" + "="*50)
print("ACCOUNT DETAILS — JOHN MENSAH")
print("="*50)
holder2.display_details()
```

## 11. Export Data to CSV
Running this cell saves three CSV files in the same folder as this notebook.

```python
all_policyholders = [holder1, holder2]
all_products      = [health_plan, auto_plan]

ph_file  = CSVExporter.export_policyholders(all_policyholders)
pr_file  = CSVExporter.export_products(all_products)
pay_file = CSVExporter.export_payments(all_policyholders)
```

## 12. Preview Exported CSVs with pandas

```python
import pandas as pd

print("── Policyholders ──")
display(pd.read_csv(ph_file))

print("\n── Products ──")
display(pd.read_csv(pr_file))

print("\n── Payments ──")
display(pd.read_csv(pay_file))
```
